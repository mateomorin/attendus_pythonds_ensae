```python
# Importation de la donnée
from sklearn.datasets import load_iris

iris, target = load_iris(as_frame=True, return_X_y=True)
iris["class"] = target
iris.head()
```

```python
# Affichage
import matplotlib.pyplot as plt

plt.scatter(iris["sepal length (cm)"], iris["sepal width (cm)"], c=iris["class"], s=40, alpha=0.8)
plt.title("Dimension des sépales des iris, colorés par classe")
plt.xlabel("Longueur du sépale (cm)")
plt.ylabel("Largeur du sépale (cm)")
plt.grid(linestyle="--", alpha=0.5)
plt.show()
```
