# Exploration des données Iris

On récupère les données Iris, essentielles à notre analyse.

## Importation

```python
from sklearn.datasets import load_iris

iris, target = load_iris(as_frame=True, return_X_y=True)
iris["class"] = target
iris.head()
```

## Premières statistiques

```python
iris.describe()
```

## Premières visualisations

```python
import matplotlib.pyplot as plt

plt.scatter(iris["sepal length (cm)"], iris["sepal width (cm)"], c=iris["class"])
plt.show()
```
